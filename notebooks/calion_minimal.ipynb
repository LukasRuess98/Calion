## Terminal Policies (Rolling Horizon):

| Policy | Beschreibung | Feasibility |
|--------|--------------|-------------|
| `equal` | SOC_end == SOC_init | Kann scheitern |
| `soft` | Weiche Nebenbedingung | **Immer lösbar** |
| `value` | Diminishing Returns Value Function | **Immer lösbar** |
| `free` | Keine Constraint | Immer lösbar |

**Aktuell konfiguriert:** `terminal_policy: value` mit diminishing returns

```python
# Setup
import sys
from pathlib import Path

current = Path.cwd()
for candidate in [current] + list(current.parents):
    if (candidate / 'calion').exists():
        if str(candidate) not in sys.path:
            sys.path.insert(0, str(candidate))
        PROJECT_ROOT = candidate
        break

from calion.run import rolling_horizon as rh
print(f"✅ Project: {PROJECT_ROOT}")
```

```python
# =============================================================================
# PRESET AUSWÄHLEN (hier ändern!)
# =============================================================================

# Option 1: Schneller Test (1 Woche, Baseline)
PRESET_QUICK = ['configs/presets/quick_test.yaml']

# Option 2: Rolling Horizon mit Value Function (EMPFOHLEN für RH-Test)
PRESET_RH_FULL = ['configs/presets/rh_full_system.yaml']

# Option 3: Wärmepumpen-Optimierung
PRESET_HP = ['configs/presets/hp_optimization.yaml']

# Option 4: Speicher-Studie
PRESET_STORAGE = ['configs/presets/storage_study.yaml']

# Option 5: Legacy (alte monolithische Config)
PRESET_LEGACY = ['configs/stadtbach.yaml']

# =============================================================================
# AKTIVES PRESET (hier ändern!)
# =============================================================================
CONFIG = PRESET_RH_FULL  # <- Für Value Function Test: PRESET_RH_FULL verwenden!

# Prüfen ob Config existiert
config_path = PROJECT_ROOT / CONFIG[0]
print(f"📋 Aktives Preset: {CONFIG[0]}")
print(f"   Existiert: {'✅' if config_path.exists() else '❌'}")

# Zeige welche Terminal Policy verwendet wird
if 'rh' in CONFIG[0].lower():
    print(f"   Terminal Policy: value (diminishing returns)")
    print(f"   → Speicher wird ökonomisch optimiert, keine harten Constraints")
```

```python
%%time
# Optimierung ausführen
print("🚀 STARTE OPTIMIERUNG\n" + "="*70)

try:
    workflow = rh.run_workflow(CONFIG)
    print("\n" + "="*70)
    print("✅ ERFOLGREICH")
    SUCCESS = True
except Exception as e:
    print(f"\n❌ FEHLER: {e}")
    import traceback
    traceback.print_exc()
    SUCCESS = False
```

```python
# Ergebnisse speichern und anzeigen
if SUCCESS:
    from calion.io.notebook_helpers import (
        display_workflow_summary, 
        display_kpi_summary,
        save_workflow_run
    )
    
    # Für Dashboard speichern
    workflow_dir = save_workflow_run(
        workflow,
        name="Minimal Test",
        description="Schnelltest",
        config_paths=CONFIG
    )
    print(f"\n📂 Gespeichert in: {workflow_dir}")
    
    display_workflow_summary(workflow)
    print("\n")
    display_kpi_summary(workflow)
else:
    print("⚠️ Keine Ergebnisse verfügbar")
```

```python
# Speicher-Check
if SUCCESS and workflow:
    result = workflow.rh_result or workflow.pf_result
    
    if 'TES_SOC_MWh' in result.series:
        import matplotlib.pyplot as plt
        import pandas as pd
        
        soc = result.series['TES_SOC_MWh']
        
        print(f"📊 Speicher SOC Analyse:")
        print(f"   Min:  {min(soc):.1f} MWh")
        print(f"   Max:  {max(soc):.1f} MWh")
        print(f"   Mean: {sum(soc)/len(soc):.1f} MWh")
        print(f"   Std:  {pd.Series(soc).std():.1f} MWh")
        
        plt.figure(figsize=(12, 4))
        plt.plot(soc, linewidth=1)
        plt.ylabel('SOC [MWh]')
        plt.title('Speicher State of Charge')
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
        
        # Check if cycling
        if pd.Series(soc).std() > 10:
            print("\n✅ Speicher cycelt (std > 10 MWh)")
        else:
            print("\n⚠️ Speicher inaktiv (std < 10 MWh)")
    else:
        print("⚠️ Keine SOC Daten verfügbar")
```
