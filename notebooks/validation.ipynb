# Stadtbach-Validierung

Dieses Notebook führt den automatischen Vergleich zwischen EnerGIS und der Legacy-Stadtbach-Referenz aus und speichert die Kennzahlen als CSV.

```python
from pathlib import Path
import sys
import os

# Auto-Setup: Projekt-Root finden
def find_project_root(start: Path) -> Path:
    """Findet das Projekt-Root-Verzeichnis."""
    for candidate in [start] + list(start.parents):
        if (candidate / '.git').exists() and (candidate / 'energis').exists():
            return candidate
    return start

PROJECT_ROOT = find_project_root(Path.cwd())
os.chdir(PROJECT_ROOT)

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print(f"✅ Projekt-Root: {PROJECT_ROOT}")

# KORRIGIERT: Absolute Pfade mit PROJECT_ROOT
from energis.validation.stadtbach import run_stadtbach_reference

REFERENCE = PROJECT_ROOT / 'tests/data/stadtbach_legacy_reference.json'
OUTPUT = PROJECT_ROOT / 'notebooks/exports/stadtbach_validation.csv'

# Sicherstellen dass Export-Verzeichnis existiert
OUTPUT.parent.mkdir(parents=True, exist_ok=True)

print(f"📄 Referenz:  {REFERENCE}")
print(f"📊 Output:    {OUTPUT}")
print(f"   Referenz existiert: {'✅' if REFERENCE.exists() else '❌'}\n")

table, run, reference = run_stadtbach_reference(reference_path=REFERENCE, output_table=OUTPUT)

table
```

Die Tabelle wird zusätzlich unter `notebooks/exports/stadtbach_validation.csv` abgelegt, so dass die CI-Pipeline sie in Artefakte übernehmen kann.
